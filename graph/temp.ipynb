```python
import numpy as np
import networkx as nx
from make import complete, barabasi
from attributes import get_eigs_sym, degrees_of, check_symmetric, is_connected
import matplotlib.pyplot as plt
from process import bye_nb, bye_st
from graph.make import erdos

from matplotlib.pyplot import cm
```

```python
g_size, n_comm, p_in, p_out = 1000, 3, 0.5, 0.001
```

```python
color=iter(cm.rainbow(np.linspace(0,1,n_comm)))
```

```python
adj, communities = erdos(g_size, n_comm, p_in, p_out)
```

```python
is_connected(adj)
```

```python
graph = nx.Graph(adj)
pos = nx.spring_layout(graph)
nx.draw(graph, pos)

for pawnee in communities:
    nx.draw_networkx_nodes(graph, pos, nodelist=list(pawnee),
                      node_color=next(color).reshape(1,-1), alpha=0.8)
```
