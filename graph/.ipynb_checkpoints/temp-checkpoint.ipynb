```python
import numpy as np
import networkx as nx
from make import complete, barabasi
from attributes import get_eigs_sym, degrees_of, check_symmetric
import matplotlib.pyplot as plt
```

```python
adj = barabasi(1000, 3)
check_symmetric(adj)
```

```python
D = np.diag(degrees_of(adj))
L = D - adj
eigs = get_eigs_sym(L)
```

```python
np.all(np.diff(eigs) >= 0)
```

```python
eigs_2 = np.sort(eigs)
```

```python
np.all(eigs == eigs_2)
```
